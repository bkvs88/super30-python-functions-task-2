Create a function that accepts a list and returns its sum without using sum().


```python
def sum_list(lst: list[int|float]) -> int|float: # defines a function named sum_list that takes a list of integers or floats as input and returns an integer or float
    '''This function calculates the sum of all elements in a list'''
    total=0 # initializes a variable total to 0, which will be used to store the sum of the elements in the list
    lst=list(lst) # converts the input lst to a list (in case it is not already a list)
    for i in lst: # iterate the list till no element found
        total= total+i  # calculate the total
    return total  # return the total sum of the list
```

```python
sum_list([1, 2, 3, 4, 5])  # calling the function with a list of numbers to calculate the sum
```

```python
sum_list([.12, 2.5, 3.1])  # calling the function with a list of float numbers to calculate the sum
```
