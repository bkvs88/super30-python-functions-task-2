Create a function using *args that accepts any number of values and returns their total.

```python
def simple_args_fun(*numbers:int|float) -> int|float: # defining the function with parameter packing as tuple
    return sum(numbers) # return the sum of given tuple as input
```

```python
simple_args_fun(1,2,3,4,6) # invoking the function with inputs as integer data type only
```

```python
simple_args_fun(1,2,3.4,5.5) #invoking the function with inputs as integer and float data type
```
