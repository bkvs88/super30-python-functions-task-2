Create a function that accepts a string and determines whether it is a palindrome.

```python
def accept_data(name,age,course):  # defining the function with 3 inputs
    ''' This function accept data and return the formatted string'''
    return f"name:{name}\n age:{age} \n course:{course}" # return the inputs
```

```python
print(accept_data("sravan",36,"Python")) # calling the function with actual sequence
```

```python
print(accept_data(age=36,course="Python",name="sravan")) # calling the function with positional arguments
```
