Create a function that accepts a string and determines whether it is a palindrome.

```python
def palindrome_check(input_str:str): # define the function name with one input string
    '''This function takes the input of string and check for given string is palindrome or not. '''
    is_palindrome=True #defining the variable to True 
    n = len(input_str) # finding the length and assigning to variable used for number of iterations.
    #print(n//2)
    for i in range(n // 2): #iterate to half of the length
        if input_str[i] != input_str[n - 1 - i]: # check if nth position from forward and backward are same.
            is_palindrome = False # in case not same. return false and exit from loop.
            break
    if is_palindrome: # using if conditional statemtnt to print the string based on the True
        return f'"{input_str}" is a palindrome string'
    else:
        return f'"{input_str}" is not a palindrome string'
```

```python
palindrome_check("racecar1")
```
