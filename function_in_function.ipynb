Create a function that accepts another function as an argument.

```python
def add(a,b): # defining the function for adding two numbers
    return a+b #returning the addition operation for the two inputs.
```

```python
def mult(num1:int,num2:int): #defining the function for multiplication
    return num1*num2 #returning the multiplication of two inputs.
```

```python
add(mult(2,3),6) #invoking the function using the invokation of another fuction output as input.
```
