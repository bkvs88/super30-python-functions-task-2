Create a function that accepts another function as an argument.

```python
def add(a: int, b: int) -> int: 
    # defining the function for adding two numbers
    return a + b 

def mult(num1: int, num2: int) -> int: 
    # defining the function for multiplication
    return num1 * num2 

def calculate(operation, x: int, y: int) -> int:
    # This is a higher-order function that accepts a function ('operation') as a parameter
    return operation(x, y)

# 🎯 Invoking the calculate function by passing the 'add' function itself (without parentheses)
addition_result = calculate(add, 10, 20)
print(f"Addition Result: {addition_result}")

# 🎯 Invoking the calculate function by passing the 'mult' function
multiplication_result = calculate(mult, 2, 3)
print(f"Multiplication Result: {multiplication_result}")

# We can also replicate your original nested logic using the higher-order function!
nested_result = calculate(add, calculate(mult, 2, 3), 6)
print(f"Nested Result: {nested_result}")
```
