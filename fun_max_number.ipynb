Create a function that returns the largest of three numbers without using max().

```python
def max_number(a,b,c): ## function named max_number that takes three parameters a, b, and c using if condition
    '''This function takes three numbers as input and returns the maximum number among them.'''
    if a >= b and a >= c: # check if a is greater than or equal to both b and c
        return a #return a string indicating that a is the maximum number
    elif b >= a and b >= c: # check if b is greater than or equal to both a and c
        return b #return a string indicating that b is the maximum number
    else: # if neither a nor b is the maximum, then c must be the maximum
        return c #return a string indicating that c is the maximum number
```

```python
max_number(21,12,33)
```

```python
def max_check_loop(a,b,c): ## function named max_check_loop that takes three parameters a, b, and c using loop
    '''This function takes three numbers as input and returns the maximum number among them using a loop.'''
    numbers = [a, b, c] # create a list of the three numbers
    max_num = numbers[0] # initialize max_num to the first number in the list
    for num in numbers: # iterate through each number in the list
        if num > max_num: # check if the current number is greater than max_num
            max_num = num # update max_num to the current number if it is greater
    return max_num #return a string indicating that max_num is the maximum number
```
