Create a mini calculator where each mathematical operation is implemented as a separate function and a main function controls the program

```python

def add(a, b): #function to define two numbers addition
    """Returns the sum of two numbers."""
    return a + b

def subtract(a, b): #substration of 2 number
    """Returns the difference of two numbers."""
    return a - b

def multiply(a, b): #multiplication of 2 number
    """Returns the product of two numbers."""
    return a * b

def divide(a, b): # to divide two numbers
    """Returns the quotient of two numbers. Handles division by zero safely."""
    if b == 0:
        return "Error: Cannot divide by zero!"
    return a / b


def main(): # main function to contolle it. 
    """Controls the user interface and program execution loop."""
    print("Welcome to the Mini Calculator!")
    
    while True:
        print("\nAvailable Operations:")
        print("1. Add (+)")
        print("2. Subtract (-)")
        print("3. Multiply (*)")
        print("4. Divide (/)")
        print("5. Exit")
        
        # Take user choice
        choice = input("Enter choice (1-5): ").strip()
        
        # Check if the user wants to exit the program
        if choice == '5':
            print("Exiting calculator. Goodbye!")
            break
            
        # Validate operation choice before asking for numbers
        if choice not in ['1', '2', '3', '4']:
            print("Invalid Input! Please select a valid option (1-5).")
            continue
            
        # Gather inputs safely with basic exception handling
        try:
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))
        except ValueError:
            print("Invalid number format! Please enter valid integers or decimals.")
            continue
            
        # Execute the chosen mathematical function
        if choice == '1':
            result = add(num1, num2)
            print(f"Result: {num1} + {num2} = {result}")
            
        elif choice == '2':
            result = subtract(num1, num2)
            print(f"Result: {num1} - {num2} = {result}")
            
        elif choice == '3':
            result = multiply(num1, num2)
            print(f"Result: {num1} * {num2} = {result}")
            
        elif choice == '4':
            result = divide(num1, num2)
            print(f"Result: {result}")
```

```python
main()
```
