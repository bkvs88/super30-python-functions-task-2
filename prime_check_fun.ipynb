Create a function that checks whether a number is prime.

```python
def prime_check(number_inp): ## function named prime_check that takes one parameter number_inp
    '''This function takes a number as input and determines if it is prime or not'''
    if number_inp < 2: # checking if the input number is less than 2, as prime numbers are greater than 1
        return f"{number_inp} is not a prime"
    else:
        for i in range(2, int(number_inp ** 0.5) + 1):# looping through each number in the range to check if it is a factor of the input number
            # print(i)
            if number_inp%i==0: # checking if the input number is divisible by the current number in the loop (i.e., if it has a factor other than 1 and itself)
                return f"{number_inp} is Not a prime" 
                break #breaking the loop if a factor is found, indicating that the number is not prime
        else: # this else block is executed if the loop completes without finding any factors, indicating that the number is prime
            return f"{number_inp} is a prime"
```
