Use lambda with filter() to extract even numbers.



```python
numbers=[1, 2, 3, 4, 5, 6] # assigning the list of numbers to variable numbers
even_number= list(filter(lambda x : x%2==0, numbers)) # lambda to square the given list. filter function takes a list from input one by one. list used to add each number output to include in list. 
```
