Use lambda with map() to square [1, 2, 3, 4, 5, 6]

```python
numbers=[1, 2, 3, 4, 5, 6] # assigning the list of numbers to variable numbers
squared= list(map(lambda x :x**2, numbers)) # lambda to square the given list. map function takes a list and iterates each number. list used to add each number output to include in list. 
```
