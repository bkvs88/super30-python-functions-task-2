Create a function that determines whether a number is even or odd.


```python
def odd_even_check(a): ## function named odd_even_check that takes one parameter a
    '''This function takes a number as input and determines if it is odd or even.'''
    if a % 2 == 0: # check if the number is divisible by 2
        return f"Given number {a} is Even" #return a string indicating that the number is even
    else: # if the number is not divisible by 2
        return f"Given number {a} is Odd" #return a string indicating that the number is odd
```

```python
odd_even_check(1)
```
