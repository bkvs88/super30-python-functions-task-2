Create a function that accepts a string and returns the number of vowels.

```python
def vowel_check(inp_string:str): # define the function for vowel check
    '''This function used to get vowel count from given string'''
    # print (f"string entered is {inp_string}") # print the input string
    inp_string=inp_string.lower() # converting input string all to lower case.
    vowels = "aeiou" #vowels variable assignment to check 
    vowel_count=0 #initiliase the variable to 0
    for char in inp_string: # loop the string for each character
        if char in vowels: # check for character in defined vowel list
            vowel_count=vowel_count+1 # if available increment the counter
    return vowel_count
```

```python
vowel_check("Hello")
```
