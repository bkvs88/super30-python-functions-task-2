Demonstrate local and global variable scope using a small program

```python
global_var=10 ## Global variable declaration any variable defining outside the function as global variable

def global_add(a): # defining the function with one parameter
    return a + global_var # using the global variable for addition
```

```python
global_add(10)
```

```python
global_var = 10 # Global variable declaration (defined outside any function)

def global_add1(a): 
    global global_var # Declaring intent to modify the global variable
    global_var = 20 # Modifying the global variable
    
    # 🎯 Explicitly defining a local variable
    local_var = 5 
    
    # Returning the sum of the parameter, the explicit local variable, and the global variable
    return a + local_var + global_var 
```

```python
# Let's test the function
result = global_add1(5)
print(f"Function Result: {result}")
print(f"Modified Global Variable: {global_var}")
```
